```python
from datetime import datetime
```

```python

class TODO:
    def __init__(self):
        self.database = []   

        print("WELCOME TO 'TO DO LIST'")

        while True:

         print('''\n Please select your operation
         Press 1 to Cheate new Task
         Press 2 to Check all Task
         Press 3 to Remove Task
         Press 4 to Exit
         ''')

         try:
           Choice=int(input("Choice: "))
           if Choice==1:
            self.create()
           elif Choice==2:
            self.check()
           elif Choice==3:
            self.remove()
           elif Choice==4:
            break
           else:
            print("Invalid choice, please try again.")

         except ValueError:
            print("Please enter a valid choice")



    def create(self):
        idx = int(len(self.database) + 1)
        mesg = input("Please enter your task:\n")
        now = datetime.now().strftime("%Y-%m-%d %H:%M:%S")  # formatted time
        self.database.append({"index":idx,"task": mesg, "time": now})
        print("Task added successfully!")

    def check(self):
        if not self.database:
            print("No tasks found.")
        else:
            print("Your tasks:")
            for item in self.database:
                print(f"- {item['index']}.  {item['task']}                           (added at {item['time']})")

    def remove(self):
         idx_no=int(input("Enter index no of completed task: "))
         for i in self.database:
            if i['index'] == idx_no:
                self.database.remove(i)

         print("Task removed successfully")


    def __data(self):
        print(self.database)
```
